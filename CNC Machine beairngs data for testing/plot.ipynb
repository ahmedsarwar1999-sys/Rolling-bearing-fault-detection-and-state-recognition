```python
plt.figure(figsize=(30, 22), dpi=300)
plt.suptitle('KDE distribution', fontsize=32, fontname='Arial')
for i, feature in enumerate(df.columns, 1):
    plt.subplot(5, 7, i)  # Adjust rows and columns based on number of features
    sns.kdeplot(df[feature], color='red', shade=True)  # Plot histogram with KDE
    plt.title(f'{feature}', fontsize=24, fontname='Arial')
    plt.xlabel('')
    plt.ylabel('')
    plt.xticks(fontsize=16)
    plt.yticks([],fontsize=16)

# Adjust layout and show the plot
plt.tight_layout()
plt.show()
```
